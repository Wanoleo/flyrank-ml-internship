```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nTrend Distribution")
print(df["trend_direction"].value_counts())

print("\nMedian Word Count by Trend")
print(df.groupby("trend_direction")["word_count"].median())
```

### Observations

- The dataset contains 30,000 webpages across multiple clients.
- Pages appear in several trend categories, indicating a mix of growing, stable, and declining content.
- Median word counts are similar across trend groups, suggesting that content length alone may not explain page performance.
